```python
identity_df = pd.read_csv("../dataset/train_identity.csv")

full_train = train_df.merge(identity_df, on="TransactionID", how="left")

full_train.head()
```

```python
missing_percentage = full_train.isnull().mean() * 100

missing_percentage.sort_values(ascending=False).head(20)
```

```python
columns_to_drop = missing_percentage[missing_percentage > 90].index

print("Columns to remove:", len(columns_to_drop))

full_train = full_train.drop(columns=columns_to_drop)
```

```python
full_train.shape
```

```python
full_train.isnull().sum().sum()
```

```python
numeric_cols = full_train.select_dtypes(include=["int64", "float64"]).columns
categorical_cols = full_train.select_dtypes(include=["object"]).columns

print("Numeric columns:", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))
```

```python
# Fill numeric columns with median
for col in numeric_cols:
    full_train[col] = full_train[col].fillna(full_train[col].median())

# Fill categorical columns with "Unknown"
for col in categorical_cols:
    full_train[col] = full_train[col].fillna("Unknown")
```

```python
full_train.isnull().sum().sum()
```

```python
categorical_cols = full_train.select_dtypes(include=["object"]).columns

print("Number of categorical columns:", len(categorical_cols))
categorical_cols
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()

for col in categorical_cols:
    full_train[col] = encoder.fit_transform(full_train[col].astype(str))
```
